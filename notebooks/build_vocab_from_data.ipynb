```python
IMAGE_DIR = ROOT_DIR/'data'/'images'
CAPTION_DIR = ROOT_DIR/'data'/'captions'
```

```python
dataset = ImageCaptionDataSet(img_dir=IMAGE_DIR, caption_dir=CAPTION_DIR)
```

```python
data_loader = DataLoader(dataset= dataset, batch_size=1)
```

```python
from src.NLP.tokenizer import Tokenizer
tokenizer = Tokenizer()
def yield_tokens():
    for _, __, caption in data_loader:
        yield tokenizer(caption[0])
```

```python
from src.NLP.build_vocab import *
```

```python
vocab = BuildVocabFromIterator(iterator=yield_tokens(), vocab_size=5000)
```
